# <div style="text-align:center; border-radius:15px 15px; padding:15px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> Linear Regression </b></div>

 
   <p>Linear regression is a crucial statistical method employed to model the relationship between a dependent variable (often referred to as the target) and one or more independent variables (often referred to as features). The goal of linear regression is to find the best-fitting linear equation that describes the relationship between these variables.

# <div style="text-align:center; border-radius:15px 15px; padding:8px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> Linear Regression Formula </b></div>

The general form of a simple linear regression equation is:</p>
        <p><code>y = θ<sub>1</sub> + θ<sub>2</sub>x + ε</code></p>
        <p>Where:</p>
        <ul>
            <li><code>y</code> is the dependent variable.</li>
            <li><code>θ<sub>1</sub></code> is the intercept of the line, the expected mean value of the dependent variable when all independent variables are zero.</li>
            <li><code>θ<sub>2</sub></code> is the slope of the line (the coefficient for the independent variable). Represent the change in the dependent variable for a one-unit change in the corresponding independent variable, holding all other variables constant.</li>
            <li><code>x</code> is the independent variable.</li>
            <li><code>ε</code> is the error term (the difference between the observed and predicted values).</li>
        </ul>
   

<a id="A"></a>
# <div style="text-align:center; border-radius:15px 15px; padding:8px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> What is it useful for? </b></div>


   <ul>
            <li><strong>Predictive Analysis:</strong> It helps in predicting the value of the dependent variable based on the independent variables.</li>
            <li><strong>Understanding Relationships:</strong> It provides insights into the strength of relationships between variables.</li>
            <li><strong>Trend Analysis:</strong> It is commonly employed in the time series analysis to understand and predict trends.</li>
            <li><strong>Feature Importance:</strong> The coefficients obtained from a linear regression model can indicate the features' importance.</li>
        </ul>

<a id="A"></a>
# <div style="text-align:center; border-radius:15px 15px; padding:8px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> Key parameters of Linear Regression </b></div>


  <ul><li><strong>R-squared (Coefficient of Determination):</strong>
                <p>Measures the proportion of the variance in the dependent variable that is predictable from the independent variables. It ranges from 0 to 1, where higher values indicate a better fit.</p>
           </li>
            <li><strong>P-values:</strong>
                <p>Indicate the statistical significance of each coefficient, where low p-value (&lt; 0.05) suggests that the independent variable is significantly related to the dependent variable.</p>
            </li>
            <li><strong>Standard Error:</strong>
                <p>Provides an estimate of the uncertainty of the coefficient estimates.</p>
            </li>
            <li><strong>Residuals:</strong>
                <p>The differences between observed and predicted values.</p>
            </li>
            <li><strong>F-statistic:</strong>
                <p>Employed to test the overall model's significance. It compares the model with no predictors (intercept only) to determine if the model with predictors is significantly better.</p>
            </li>
            <li><strong>Assumptions:</strong>
                <p>Linear regression assumes linearity, independence, homoscedasticity (constant variance of errors), and normality of residuals. Violation of these assumptions could affect the validity of the model.</p>
            </li>
        </ul>

# <div style="text-align:center; border-radius:15px 15px; padding:15px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> Regularization Methods </b></div>

In Machine Learning, the optimal algorithm should exhibit low bias and must effectively approximate the actual relationship. Two frequently employed techniques to strike a balance between simple and complex models are L1 and L2 regularization.

<a id="A"></a>
# <div style="text-align:center; border-radius:15px 15px; padding:15px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> L2 Ridge Regression </b></div>

L2 Ridge Regression is a regularization technique to mitigate overfitting. It reduces the high variance seen in ordinary least squares (OLS) by introducing a bias. Ridge regression, or Tikhonov regularization, addresses multicollinearity in multiple regression data. While OLS estimates are unbiased but have high variance, ridge regression adds a penalty to the coefficient sizes, shrinking them and thus controlling overfitting.

<a id="A"></a>
# <div style="text-align:center; border-radius:15px 15px; padding:8px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> L2 Ridge Regression Formula</b></div>

<p class="formula">minimize &sum;(y<sub>i</sub> - β<sub>0</sub> - &sum;β<sub>j</sub>x<sub>ij</sub>)<sup>2</sup> + λ &sum;β<sub>j</sub><sup>2</sup></p>
        <p>Where:</p>
        <ul>
            <li><code>y<sub>i</sub></code> is the dependent variable.</li>
            <li><code>x<sub>ij</sub></code> are the independent variables.</li>
            <li><code>β<sub>0</sub></code> is the intercept.</li>
            <li><code>β<sub>j</sub></code> are the coefficients.</li>
            <li><code>λ</code> is the regularization parameter (penalty term).</li>
        </ul>
        <p>The penalty term <code>λ &sum;β<sub>j</sub><sup>2</sup></code> discourages large coefficients, effectively shrinking them towards zero.</p>

 <div>
        <h1>Regularization Penalties</h1>
        <p>The L2 penalty is quadratic: <code>lambda * slope^2</code>, preventing large coefficients.</p>
        <p>The L1 penalty is the absolute value: <code>lambda * |slope|</code>, selecting key features.</p>
        <p>When <code>lambda = 0</code>, there's no penalty, minimizing the sum of squared residuals.</p>
        <p>As <code>lambda</code> increases, the slope approaches zero, making predictions less sensitive to independent variables.</p>
        <p>Cross-validation, often 10-fold, determines the best <code>lambda</code> to minimize variance.</p>
        <p><code>Lambda</code> is the tuning parameter controlling the bias-variance tradeoff, optimized via cross-validation.</p>
    </div>

# <div style="text-align:center; border-radius:15px 15px; padding:15px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> L1 Lasso Regression </b></div>

L1 Lasso Regression is a technique used for regularization to prevent overfitting. Although it is akin to Ridge Regression, it differs significantly in its penalty function.

<a id="A"></a>
# <div style="text-align:center; border-radius:15px 15px; padding:8px; color:white; margin:0; font-size:100%; font-family:Pacifico; background-color:RoyalBlue; overflow:hidden"><b> L1 Lasso Regression Formula</b></div>

 <p class="formula">minimize &sum;(y<sub>i</sub> - β<sub>0</sub> - &sum;β<sub>j</sub>x<sub>ij</sub>)<sup>2</sup> + λ &sum;|β<sub>j</sub>|</p>
        <p>Where:</p>
        <ul>
            <li><code>y<sub>i</sub></code> is the dependent variable.</li>
            <li><code>x<sub>ij</sub></code> are the independent variables.</li>
            <li><code>β<sub>0</sub></code> is the intercept.</li>
            <li><code>β<sub>j</sub></code> are the coefficients.</li>
            <li><code>λ</code> is the regularization parameter (penalty term).</li>
        </ul>
        <p>The penalty term <code>λ &sum;|β<sub>j</sub>|</code> encourages some coefficients to be exactly zero, effectively performing variable selection.</p>

<div style="text-align: center;">
    <img src="https://www.andreaperlato.com/img/lassoregression.png" alt="alzheimers" style="width:100%; height:auto; border-radius: 30px;">
</div>

The results of Lasso Regression are similar to Ridge Regression and can be applied in logistic regression, discrete value regression, and interaction regression. The key difference arises when increasing the value of lambda. Ridge Regression can only shrink coefficients close to zero, while Lasso Regression can shrink them to exactly zero. This is beneficial when dealing with many parameters, as Lasso can eliminate insignificant ones by setting their coefficients to zero, whereas Ridge may still keep them at non-zero values.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix
corr_matrix = df1.corr(method="pearson")

# Set up the matplotlib figure
plt.figure(figsize=(12, 12))

# Draw the heatmap with the mask and correct aspect ratio
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='Spectral', cbar_kws={'shrink': .8}, 
            linewidths=0.5, linecolor='black', annot_kws={"size": 10})

# Customize the plot
plt.xticks(rotation=45, fontsize=12)
plt.yticks(rotation=0, fontsize=12)
plt.title('Correlation Matrix of Selected Features', fontsize=16)
plt.show()
```

```python
df2=corr_matrix.index[abs(corr_matrix['price'])>=0.3].tolist()
df2.remove('price')   
print(df2)
```

```python
y=df['price']
x=df[df2]
scaler = StandardScaler() 

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.1)
X_train = scaler.fit_transform(x_train)
X_test = scaler.transform(x_test)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
linear_regression=LinearRegression()
linear_regression.fit(x_train,y_train)
y_test_pred_lr=linear_regression.predict(x_test)
linear_regression_acc=r2_score(y_test,y_test_pred_lr)
print(f'Linear Regression Test accuracy: {linear_regression_acc}')
```

```python

residuals = y_test - y_test_pred_lr

plt.figure(figsize=(10, 6))
sns.histplot(residuals, kde=True, color='blue', bins=30, label='Residuals')
plt.xlabel('Residuals')
plt.ylabel('Frequency')
plt.title('Distribution of Residuals for Linear Regression Model')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(y_test_pred_lr, residuals, alpha=0.5, label='Residuals')
plt.axhline(y=0, color='red', linestyle='--', linewidth=2, label='Zero Error')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residuals vs Predicted Values for Linear Regression Model')
plt.legend()
plt.grid(True)
plt.show()
```

```python
ridge = Ridge(alpha=0.5)
ridge.fit(x_train, y_train)
y_test_pred_ridge = ridge.predict(x_test)
ridge_acc = r2_score(y_test, y_test_pred_ridge)
print(f'Ridge Regression Test accuracy: {ridge_acc}')
```

```python
lasso = Lasso(alpha=0.5)
lasso.fit(x_train, y_train)
y_test_pred_lasso = lasso.predict(x_test)
lasso_acc = r2_score(y_test, y_test_pred_lasso)
print(f'Lasso Regression Test accuracy: {lasso_acc}')
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# First subplot: Linear Regression
axes[0].scatter(y_test, y_test_pred_lr, alpha=0.5, label='Predicted vs Actual')
axes[0].plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--', linewidth=2, label='Perfect Fit')
axes[0].set_xlabel('Actual Values')
axes[0].set_ylabel('Predicted Values')
axes[0].set_title('Predictions vs Actual Values for Linear Regression Model')
axes[0].legend()
axes[0].grid(True)

# Second subplot: Ridge Regression
axes[1].scatter(y_test, y_test_pred_ridge, alpha=0.5, label='Predicted vs Actual')
axes[1].plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--', linewidth=2, label='Perfect Fit')
axes[1].set_xlabel('Actual Values')
axes[1].set_ylabel('Predicted Values')
axes[1].set_title('Predictions vs Actual Values for Ridge Regression Model')
axes[1].legend()
axes[1].grid(True)

# Third subplot: Lasso Regression
axes[2].scatter(y_test, y_test_pred_lasso, alpha=0.5, label='Predicted vs Actual')
axes[2].plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--', linewidth=2, label='Perfect Fit')
axes[2].set_xlabel('Actual Values')
axes[2].set_ylabel('Predicted Values')
axes[2].set_title('Predictions vs Actual Values for Lasso Regression Model')
axes[2].legend()
axes[2].grid(True)

axes[2].text(0.5, 0.5, 0.5, fontsize=20, ha='center') # Placeholder text

# Adjust layout to avoid overlap
plt.tight_layout()
plt.show()
```
